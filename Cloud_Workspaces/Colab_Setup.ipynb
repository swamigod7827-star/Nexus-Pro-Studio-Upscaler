```python
# 5. Start Server and LocalTunnel
import time
import threading
import sys
import subprocess

print('\n\n🔥 STARTING NEXUS PRO ON COLAB GPU 🔥')
flask_p = subprocess.Popen(['python', 'cloud_worker.py'], stdout=subprocess.PIPE, stderr=subprocess.STDOUT)

def print_output(process):
    for line in iter(process.stdout.readline, b''):
        sys.stdout.write(line.decode('utf-8'))

threading.Thread(target=print_output, args=(flask_p,)).start()
time.sleep(5)

p = subprocess.Popen(['lt', '--port', str(PORT)], stdout=subprocess.PIPE)
out = p.stdout.readline().decode('utf-8').strip()
url = out.split(' ')[-1]
print(f'\n\n👉 COPY THIS URL INTO YOUR LOCAL APP: {url} 👈')
print('=' * 50 + '\n')

p.wait()
```
